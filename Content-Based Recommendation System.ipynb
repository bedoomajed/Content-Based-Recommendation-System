```python
import pandas as pd

movies = pd.read_csv("movies_dataset.csv")

print(movies.head())
```

```python
def calculate_score(movie_genres, user_preferences):

    score = 0

    movie_genres = movie_genres.split()

    for genre in user_preferences:

        if genre in movie_genres:
            score += 1

    return score
```

```python
user_input = input("Enter favorite genres: ")

user_preferences = user_input.lower().split()
```

```python
movies["score"] = movies["genre"].apply(
    lambda x: calculate_score(x, user_preferences)
)

recommendations = movies.sort_values(
    by="score",
    ascending=False
)

print(recommendations[["title", "score"]])
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
vectorizer = CountVectorizer()

matrix = vectorizer.fit_transform(
    movies["genre"]
)
```

```python
similarity = cosine_similarity(matrix)
```

```python
movie_name = "Interstellar"
```

```python
index = movies[movies["title"] == movie_name].index[0]

scores = list(enumerate(similarity[index]))

sorted_scores = sorted(
    scores,
    key=lambda x: x[1],
    reverse=True
)
```

```python
for movie in sorted_scores[1:6]:

    print(movies.iloc[movie[0]]["title"])
```
